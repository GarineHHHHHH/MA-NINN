```python
import torch
import torch.nn as nn

class MultiHeadAttention(nn.Module):
    def __init__(self, input_dim, d_k, d_v, h, output_dim):
        super(MultiHeadAttention, self).__init__()
        self.input_dim = input_dim
        self.d_k = d_k
        self.d_v = d_v
        self.h = h
        self.output_dim = output_dim

        self.W_Q = nn.Linear(input_dim, d_k)
        self.W_K = nn.Linear(input_dim, d_k)
        self.W_V = nn.Linear(input_dim, d_v)
        self.W_O = nn.Linear(d_v, output_dim)

    def forward(self, X):
        # 线性变换
        Q = self.W_Q(X)  # (60 * d_k)
        K = self.W_K(X)  # (60 * d_k)
        V = self.W_V(X)  # (60 * d_v)

        # 多头划分
        Q_heads = Q.chunk(self.h, dim=-1)  # 长度为 h 的列表，每个元素为 (60 * (d_k / h))
        K_heads = K.chunk(self.h, dim=-1)  # 长度为 h 的列表，每个元素为 (60 * (d_k / h))
        V_heads = V.chunk(self.h, dim=-1)  # 长度为 h 的列表，每个元素为 (60 * (d_v / h))

        # 计算每个头的注意力得分
        head_outputs = []
        for Q_head, K_head, V_head in zip(Q_heads, K_heads, V_heads):
            attention_scores = torch.matmul(Q_head, K_head.transpose(-2, -1)) / torch.sqrt(torch.tensor(self.d_k / self.h, dtype=torch.float32))  # (60 * 60)
            attention_weights = torch.softmax(attention_scores, dim=-1)  # (60 * 60)
            head_output = torch.matmul(attention_weights, V_head)  # (60 * (d_v / h))
            head_outputs.append(head_output)

        # 拼接多头输出
        concat_output = torch.cat(head_outputs, dim=-1)  # (60 * d_v)

        # 最终线性变换
        output = self.W_O(concat_output)  # (60 * output_dim)

        return output

# 示例使用
input_dim = 32
d_k = 64
d_v = 64
h = 8
output_dim = 32

multihead_attn = MultiHeadAttention(input_dim, d_k, d_v, h, output_dim)
X = torch.randn(60, 32)  # (60 * 32)
output = multihead_attn(X)
print(output.shape)  # 输出: (60 * 32)
```
